# Earthquake EDA
Bu notebook Gold Delta tablolarını hızlıca incelemek için hazırlanmıştır.


```python
import os
from pyspark.sql import SparkSession
from delta import configure_spark_with_delta_pip

builder = (SparkSession.builder.appName('earthquake_eda')
    .master('local[*]')
    .config('spark.sql.extensions', 'io.delta.sql.DeltaSparkSessionExtension')
    .config('spark.sql.catalog.spark_catalog', 'org.apache.spark.sql.delta.catalog.DeltaCatalog'))
spark = configure_spark_with_delta_pip(builder).getOrCreate()
```

```python
features_path = '/app/data/delta/gold/earthquake_features'
df = spark.read.format('delta').load(features_path)
df.printSchema()
df.show(5, truncate=False)
```

```python
df.groupBy('year').count().orderBy('year').show(40)
df.groupBy('state').count().orderBy('count', ascending=False).show(20, truncate=False)
```
