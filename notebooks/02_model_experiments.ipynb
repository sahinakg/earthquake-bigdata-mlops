# Model Experiments
MLflow deneylerini programatik olarak listelemek için örnek notebook.


```python
import mlflow
from mlflow import MlflowClient

mlflow.set_tracking_uri('http://mlflow:5000')
client = MlflowClient()
for exp in client.search_experiments():
    print(exp.name, exp.experiment_id)
```

```python
experiment = client.get_experiment_by_name('earthquake_magnitude_regression')
if experiment:
    runs = client.search_runs([experiment.experiment_id], order_by=['metrics.rmse ASC'], max_results=5)
    for run in runs:
        print(run.info.run_id, run.data.metrics, run.data.params.get('model_name'))
```
